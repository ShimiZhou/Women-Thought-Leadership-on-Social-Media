```python
# API call
import requests
import pandas as pd

root = "https://ensembledata.com/apis"
endpoint = "/youtube/search"
params = {
  "keyword": "ai",
  "depth": 5,
  "start_cursor": "",
  "period": "overall",
  "sorting": "relevance",
  "get_additional_info": False,
  "token": "vFA99y6eg5VWTCbH"
}

res = requests.get(root+endpoint, params=params)
print(res.json())
data_leadership = res.json()
```

```python
# Step 1: Get video entries
import pandas as pd
posts = data_leadership['data']['posts']

# Step 2: Extract relevant metadata
videos = []
for post in posts:
    vr = post.get("videoRenderer", {})
    title = vr.get("title", {}).get("runs", [{}])[0].get("text", "")
    channel = vr.get("longBylineText", {}).get("runs", [{}])[0].get("text", "")
    published = vr.get("publishedTimeText", {}).get("simpleText", "")
    views_text = vr.get("viewCountText", {}).get("simpleText", "0 views").split()[0].replace(",", "").replace("K", "000").replace("M", "000000")
    try:
        views = int(views_text)
    except:
        views = None
    description = ""
    if "detailedMetadataSnippets" in vr:
        runs = vr["detailedMetadataSnippets"][0]["snippetText"]["runs"]
        description = " ".join([r.get("text", "") for r in runs])
    video_id = vr.get("videoId", "")
    video_url = f"https://www.youtube.com/watch?v={video_id}"

    videos.append({
        "title": title,
        "description": description,
        "channel": channel,
        "published": published,
        "views": views,
        "video_url": video_url
    })
```

```python
# Step 3: Convert to DataFrame
df = pd.DataFrame(videos)
print(df.head())
```

```python
# Save DataFrame to Excel
df.to_excel("youtube_ai_videos.xlsx", index=False)
print(f"Saved {len(df)} videos to 'youtube_ai_videos.xlsx'")
```

```python
# channel, views, and frequency of appearance

# Define “thought leader” as: 
# mean: High average views/likes/comments per video; 
# max: High subscriber count (if accessible)
# count: Frequent appearance in top results

leader1_stats = df.groupby('channel').agg({
    'views': ['mean', 'max', 'count']
}).sort_values(('views', 'mean'), ascending=False)

leader1_stats
```
